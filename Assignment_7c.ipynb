Likelihood Contribution of a Single Response $y_k$

Let $p_k(\theta) = P(Y_k = 1 \mid \Theta = \theta) = \frac{1}{1 + e^{-a_k(\theta - b_k)}}$ be the probability of answering item $k$ correctly.Since $y_k \in \{0, 1\}$ represents a Bernoulli trial conditionally independent on $\theta$, the likelihood contribution of the single response $y_k$ is written as:$$L(y_k \mid \theta) = [p_k(\theta)]^{y_k} [1 - p_k(\theta)]^{1 - y_k}$$Alternatively, using the logistic function definition explicitly:$$L(y_k \mid \theta) = \left( \frac{1}{1 + e^{-a_k(\theta - b_k)}} \right)^{y_k} \left( \frac{e^{-a_k(\theta - b_k)}}{1 + e^{-a_k(\theta - b_k)}} \right)^{1 - y_k}$$2. Joint Likelihood Function for the Running History $\mathbf{y}^{(k)}$Assuming responses are conditionally independent given the user's latent ability $\theta$, the joint likelihood function for the running history vector $\mathbf{y}^{(k)} = (y_1, y_2, \dots, y_k)$ is the product of individual likelihood contributions up to step $k$:$$L(\mathbf{y}^{(k)} \mid \theta) = \prod_{i=1}^{k} L(y_i \mid \theta) = \prod_{i=1}^{k} [p_i(\theta)]^{y_i} [1 - p_i(\theta)]^{1 - y_i}$$

By Bayes' theorem, the posterior distribution at step $k$, given the updated response history vector $\mathbf{y}^{(k)} = (\mathbf{y}^{(k-1)}, y_k)$, is proportional to the product of the single-step likelihood contribution $L(y_k \mid \theta)$ and the posterior distribution from the previous step $k-1$:$$f_{\Theta \mid \mathbf{Y}^{(k)}}(\theta \mid \mathbf{y}^{(k)}) \propto L(y_k \mid \theta) \cdot f_{\Theta \mid \mathbf{Y}^{(k-1)}}(\theta \mid \mathbf{y}^{(k-1)})$$Expanded FormSubstituting $L(y_k \mid \theta) = [p_k(\theta)]^{y_k} [1 - p_k(\theta)]^{1 - y_k}$, we get:$$f_{\Theta \mid \mathbf{Y}^{(k)}}(\theta \mid \mathbf{y}^{(k)}) \propto [p_k(\theta)]^{y_k} [1 - p_k(\theta)]^{1 - y_k} \cdot f_{\Theta \mid \mathbf{Y}^{(k-1)}}(\theta \mid \mathbf{y}^{(k-1)})$$where:$f_{\Theta \mid \mathbf{Y}^{(k-1)}}(\theta \mid \mathbf{y}^{(k-1)})$ acts as the prior density for step $k$ (with $f_{\Theta \mid \mathbf{Y}^{(0)}}(\theta) = f_{\Theta}^{(0)}(\theta)$ being the initial standard normal prior $\mathcal{N}(0,1)$).$[p_k(\theta)]^{y_k} [1 - p_k(\theta)]^{1 - y_k}$ is the likelihood contribution of the newly observed response $y_k$.

When a user answers a question correctly ($y_k = 1$), the likelihood contribution simplifies to:$$L(y_k = 1 \mid \theta) = p_k(\theta) = \frac{1}{1 + e^{-a_k(\theta - b_k)}}$$This function $p_k(\theta)$ is a monotonically increasing S-curve (sigmoid) bounded between $0$ and $1$

:As $\theta \to -\infty$, $p_k(\theta)
 \to 0$.As $\theta \to \infty$, $p_k(\theta) \to
1$

.Effect of a Large Difficulty Parameter ($b_k$)Rightward Shift of the Likelihood Curve:When the item is highly difficult (large positive $b_k$), the inflection point of $p_k(\theta)$ (where $p_k(\theta) = 0.5$) is pushed far to the right.Steep Weighting Gradient at Normal Ability Levels:For typical or lower ability levels ($\theta < b_k$), the likelihood function $p_k(\theta)$ is small and increases rapidly with respect to $\theta$. Because $p_k(\theta)$ acts as a multiplicative weighting factor on the previous posterior $f_{\Theta \mid \mathbf{Y}^{(k-1)}}(\theta \mid \mathbf{y}^{(k-1)})$, it heavily suppresses (dampens) lower values of $\theta$ while favoring higher values.Shift in the Mode (MAP Estimate):To find the peak (mode/Maximum A Posteriori estimate) of the new posterior density, we look at the log-posterior derivative:$$\frac{d}{d\theta} \ln f_{\Theta \mid \mathbf{Y}^{(k)}}(\theta \mid \mathbf{y}^{(k)}) = \frac{d}{d\theta} \ln f_{\Theta \mid \mathbf{Y}^{(k-1)}}(\theta \mid \mathbf{y}^{(k-1)}) + \frac{d}{d\theta} \ln p_k(\theta)$$The gradient contribution from the correct answer is:$$\frac{d}{d\theta} \ln p_k(\theta) = a_k [1 - p_k(\theta)] > 0$$Since $b_k$ is large relative to the current mode $\hat{\theta}_{k-1}$, $p_k(\hat{\theta}_{k-1})$ is small, making $1 - p_k(\hat{\theta}_{k-1})$ large.This adds a strong positive slope to the log-posterior at the previous peak location $\hat{\theta}_{k-1}$.Consequently, the derivative at $\hat{\theta}_{k-1}$ becomes strictly positive, forcing the new peak $\hat{\theta}_k$ to shift significantly to the right (towards higher ability $\theta$).

The variance (uncertainty) of the posterior distribution is inversely proportional to its curvature at the peak (mode). Using a Gaussian approximation (Laplace approximation), the posterior precision (inverse variance) at step $k$ satisfies:$$\text{Precision}_k = \frac{1}{\sigma_k^2} \approx \frac{1}{\sigma_{k-1}^2} + I_k(\theta)$$where $I_k(\theta)$ is the Item Information Function contributed by observation $k$:$$I_k(\theta) = a_k^2 \cdot p_k(\theta) [1 - p_k(\theta)]$$2. Role of Discrimination ($a_k$) in Posterior SharpnessBecause $I_k(\theta)$ scales quadratically with $a_k$ (i.e., $\propto a_k^2$):Higher $a_k$ yields higher information gain, adding greater curvature (negative second derivative) to the log-posterior.Higher curvature results in a sharper, narrower peak, directly reducing the variance ($\sigma^2$) and increasing model certainty about the user's latent ability $\theta$.3. Comparing Extreme Cases ($a_k$ Very Large vs. Very Small)Case A: $a_k$ is Very Large ($a_k \gg 1$)Mechanics: The item characteristic curve (ICC) becomes nearly a step function (Heaviside step) centered at difficulty $b_k$.Information Gain: Around $\theta \approx b_k$, the item provides extreme information ($I_k(\theta) \propto a_k^2$).Posterior Behavior:The posterior updates drastically, yielding a very sharp peak (low variance).It provides near-certainty about whether $\theta > b_k$ (if $y_k=1$) or $\theta < b_k$ (if $y_k=0$), cutting off the rejected region of ability space sharply.Case B: $a_k$ is Very Small ($a_k \approx 0$)Mechanics: The probability curve $p_k(\theta) \approx 0.5$ becomes nearly flat across all ability levels $\theta$.Information Gain: Information $I_k(\theta) \to 0$ regardless of where the user's ability lies.Posterior Behavior:The likelihood function is virtually constant, meaning the observation $y_k$ contains almost no signal about $\theta$.The posterior distribution remains nearly identical to the prior state ($f_{\Theta \mid \mathbf{Y}^{(k)}} \approx f_{\Theta \mid \mathbf{Y}^{(k-1)}}$).No reduction in variance occurs; uncertainty remains unchanged.

1. Algorithmic Framework (Grid Approximation)Step 1: Discretize the Latent Ability Space ($\Theta$)Define a bounded interval for ability, e.g., $\theta \in [-4, 4]$.Discretize this continuous interval into $M$ equally spaced evaluation points:$$\boldsymbol{\theta} = [\theta_1, \theta_2, \dots, \theta_M]$$Calculate the constant grid spacing:$$\Delta\theta = \frac{\theta_M - \theta_1}{M - 1}$$Step 2: Initialize Prior Density ($k = 0$)Evaluate the initial standard normal prior $f_\Theta^{(0)}(\theta) = \frac{1}{\sqrt{2\pi}} \exp\left(-\frac{\theta^2}{2}\right)$ over all grid points $m \in \{1, 2, \dots, M\}$:$$p_m^{(0)} = f_\Theta^{(0)}(\theta_m)$$Normalize initial prior using the trapezoidal/Riemann sum so that $\sum_{m=1}^{M} p_m^{(0)} \Delta\theta = 1$:$$\mathbf{f}^{(0)} = \frac{\mathbf{p}^{(0)}}{\sum_{m=1}^{M} p_m^{(0)} \Delta\theta}$$Step 3: Sequential Bayesian Update Loop (Step $k$)When a user responds to item $k$ with response $y_k \in \{0, 1\}$, given item parameters $a_k$ and $b_k$:Compute Pointwise Likelihood Vector:For each grid point $\theta_m$, evaluate the 2PL success probability:$$P(Y_k = 1 \mid \Theta = \theta_m) = \pi_k(\theta_m) = \frac{1}{1 + e^{-a_k(\theta_m - b_k)}}$$Construct the likelihood vector $\mathbf{L}_k = [L_1, L_2, \dots, L_M]^T$:$$L_m = [\pi_k(\theta_m)]^{y_k} [1 - \pi_k(\theta_m)]^{1 - y_k}$$Compute Unnormalized Posterior Vector:Perform an element-wise (Hadamard) multiplication of the likelihood vector $\mathbf{L}_k$ and the prior vector $\mathbf{f}^{(k-1)}$:$$\tilde{f}_m^{(k)} = L_m \cdot f_m^{(k-1)} \quad \text{for } m = 1, 2, \dots, M$$2. Sequential Normalization StepTo convert the unnormalized values $\tilde{\mathbf{f}}^{(k)}$ back into a valid continuous probability density function ($f_{\Theta \mid \mathbf{Y}^{(k)}}(\theta_m \mid \mathbf{y}^{(k)})$), perform numerical integration via Trapezoidal Rule or Riemann Sum:Computational FormulaCompute Marginal Likelihood (Evidence / Normalization Constant):$$C_k = \sum_{m=1}^{M} \tilde{f}_m^{(k)} \, \Delta\theta$$(Note: Using the Trapezoidal rule gives $C_k = \Delta\theta \left( \frac{\tilde{f}_1^{(k)} + \tilde{f}_M^{(k)}}{2} + \sum_{m=2}^{M-1} \tilde{f}_m^{(k)} \right)$ for even higher accuracy).Normalize Density Vector:Divide each element of the unnormalized vector by $C_k$:$$f_m^{(k)} = \frac{\tilde{f}_m^{(k)}}{C_k} \quad \text{for } m = 1, 2, \dots, M$$Numerical Stability Note: To avoid underflow errors when $n$ is large, computations can be performed in log-space ($\ln L_m + \ln f_m^{(k-1)}$) using the log-sum-exp trick before exponentiating back to obtain the normalized vector.

```python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Set random seed for reproducibility
np.random.seed(42)

# --- 1. Simulation Parameters ---
theta_true = 0.75
n_items = 20

# Item parameters for n = 20 items (randomly generated realistic IRT parameters)
# Discrimination a_i in [0.8, 2.5], Difficulty b_i in [-2, 2]
a_params = np.random.uniform(0.8, 2.5, size=n_items)
b_params = np.random.uniform(-2.0, 2.0, size=n_items)


# --- 2. Grid Discretization Setup ---
M = 500
theta_grid = np.linspace(-4, 4, M)
d_theta = theta_grid[1] - theta_grid[0]

# Standard Normal Prior Initial Distribution: N(0, 1)
prior_density = (1 / np.sqrt(2 * np.pi)) * np.exp(-(theta_grid**2) / 2)
prior_density /= np.sum(prior_density * d_theta)  # Ensure integration to 1


# --- 3. Helper Functions ---
def irt_2pl_prob(theta, a, b):
  """Probability of correct response under 2PL IRT model."""
  return 1.0 / (1.0 + np.exp(-a * (theta - b)))


def get_credible_interval(grid, density, d_theta, alpha=0.05):
  """Computes (1 - alpha) Equal-Tailed Credible Interval."""
  cdf = np.cumsum(density * d_theta)
  lower_bound = grid[np.searchsorted(cdf, alpha / 2)]
  upper_bound = grid[np.searchsorted(cdf, 1 - alpha / 2)]
  return lower_bound, upper_bound


# --- 4. Running Simulation Loop ---
running_posterior = prior_density.copy()

# Data containers to track over steps 0 to n
eap_estimates = []
map_estimates = []
ci_lower = []
ci_upper = []
responses = []

# Record Initial Prior State (Step k = 0)
eap_0 = np.sum(theta_grid * running_posterior * d_theta)
map_0 = theta_grid[np.argmax(running_posterior)]
lower_0, upper_0 = get_credible_interval(
    theta_grid, running_posterior, d_theta
)

eap_estimates.append(eap_0)
map_estimates.append(map_0)
ci_lower.append(lower_0)
ci_upper.append(upper_0)

# Iterate through items
for k in range(n_items):
  a_k = a_params[k]
  b_k = b_params[k]

  # Generate User Response based on True Ability
  p_success = irt_2pl_prob(theta_true, a_k, b_k)
  y_k = 1 if np.random.rand() < p_success else 0
  responses.append(y_k)

  # Likelihood vector across the grid
  p_grid = irt_2pl_prob(theta_grid, a_k, b_k)
  likelihood = (p_grid**y_k) * ((1.0 - p_grid) ** (1 - y_k))

  # Bayesian Update & Normalization
  unnorm_posterior = likelihood * running_posterior
  running_posterior = unnorm_posterior / np.sum(unnorm_posterior * d_theta)

  # Calculate Estimators:
  # EAP (Expected A Posteriori): Posterior Mean
  eap_k = np.sum(theta_grid * running_posterior * d_theta)
  # MAP (Maximum A Posteriori): Posterior Mode
  map_k = theta_grid[np.argmax(running_posterior)]

  # Credible Interval (95%)
  low_k, high_k = get_credible_interval(theta_grid, running_posterior, d_theta)

  # Store metrics
  eap_estimates.append(eap_k)
  map_estimates.append(map_k)
  ci_lower.append(low_k)
  ci_upper.append(high_k)

# --- 5. Visualization with Plotly ---
steps = np.arange(n_items + 1)

fig = go.Figure()

# 95% Credible Interval (Shaded Area)
fig.add_trace(
    go.Scatter(
        x=np.concatenate([steps, steps[::-1]]),
        y=np.concatenate([ci_upper, ci_lower[::-1]]),
        fill='toself',
        fillcolor='rgba(31, 119, 180, 0.2)',
        line=dict(color='rgba(255,255,255,0)'),
        hoverinfo='skip',
        showlegend=True,
        name='95% Credible Interval',
    )
)

# EAP Estimate Line
fig.add_trace(
    go.Scatter(
        x=steps,
        y=eap_estimates,
        mode='lines+markers',
        name='EAP (Posterior Mean)',
        line=dict(color='#1f77b4', width=2.5),
        marker=dict(size=6),
    )
)

# MAP Estimate Line
fig.add_trace(
    go.Scatter(
        x=steps,
        y=map_estimates,
        mode='lines+markers',
        name='MAP (Posterior Mode)',
        line=dict(color='#ff7f0e', width=2, dash='dash'),
        marker=dict(size=5),
    )
)

# True Theta Line
fig.add_trace(
    go.Scatter(
        x=[0, n_items],
        y=[theta_true, theta_true],
        mode='lines',
        name=f'True Ability (θ = {theta_true})',
        line=dict(color='red', width=2, dash='dot'),
    )
)

# Formatting
fig.update_layout(
    title=dict(
        text=(
            '<b>Convergence of Ability Estimators over Time (n = 20)</b><br><sup>'  # noqa: E501
            f'True Ability θ = {theta_true} | Tracking EAP, MAP, and 95%'
            ' Credible Bounds</sup>'
        ),
        x=0.5,
    ),
    xaxis=dict(title='Item Sequence Step (k)', dtick=1),
    yaxis=dict(title=r'Latent Ability Estimate ($\theta$)', range=[-2.5, 2.5]),
    template='plotly_white',
    width=900,
    height=550,
    legend=dict(x=0.02, y=0.98, bordercolor='Black', borderwidth=1),
)

fig.show()
```

```python
import numpy as np
import plotly.graph_objects as go

# Set random seed for reproducible results
np.random.seed(42)

# --- 1. Simulation Parameters ---
theta_true = 0.75
n_items = 20

# Item Parameter Distributions specified in the prompt:
# b_k ~ N(0, 1)
# a_k ~ Uniform(0.5, 2.0)
b_params = np.random.normal(0, 1, size=n_items)
a_params = np.random.uniform(0.5, 2.0, size=n_items)

# --- 2. Grid Discretization Setup ---
M = 500  # Number of grid points
theta_grid = np.linspace(-4, 4, M)
d_theta = theta_grid[1] - theta_grid[0]

# Standard Normal Prior f_0(theta) ~ N(0, 1)
prior_density = (1 / np.sqrt(2 * np.pi)) * np.exp(-(theta_grid**2) / 2)
prior_density /= np.sum(prior_density * d_theta)  # Normalize prior


# --- 3. Helper 2PL Function ---
def irt_2pl_prob(theta, a, b):
  return 1.0 / (1.0 + np.exp(-a * (theta - b)))


# --- 4. Sequential Tracking Setup ---
running_posterior = prior_density.copy()

# Trackers from Step 0 to 20
theta_bayes_list = []  # Posterior Mean
theta_map_list = []  # MAP (Mode)

# Step 0: Prior Metrics
initial_bayes = np.sum(theta_grid * running_posterior * d_theta)
initial_map = theta_grid[np.argmax(running_posterior)]

theta_bayes_list.append(initial_bayes)
theta_map_list.append(initial_map)

# --- 5. Running Update Loop ---
for k in range(n_items):
  a_k = a_params[k]
  b_k = b_params[k]

  # Response generation: y_k = 1 if U(0, 1) < p_k(theta_true) else 0
  p_true = irt_2pl_prob(theta_true, a_k, b_k)
  y_k = 1 if np.random.uniform(0, 1) < p_true else 0

  # Pointwise likelihood across theta grid
  p_grid = irt_2pl_prob(theta_grid, a_k, b_k)
  likelihood = (p_grid**y_k) * ((1.0 - p_grid) ** (1 - y_k))

  # Bayesian Update & Normalization
  unnorm_posterior = likelihood * running_posterior
  running_posterior = unnorm_posterior / np.sum(unnorm_posterior * d_theta)

  # Calculate Estimators at Step k
  eap_k = np.sum(theta_grid * running_posterior * d_theta)  # Posterior Mean
  map_k = theta_grid[np.argmax(running_posterior)]  # MAP Estimate

  theta_bayes_list.append(eap_k)
  theta_map_list.append(map_k)

# --- 6. Visualization using Plotly ---
steps = np.arange(n_items + 1)

fig = go.Figure()

# Running Posterior Mean (Theta_Bayes)
fig.add_trace(
    go.Scatter(
        x=steps,
        y=theta_bayes_list,
        mode="lines+markers",
        name=r"$\text{Posterior Mean } (\hat{\theta}_{\text{Bayes}}^{(k)})$",
        line=dict(color="#1f77b4", width=2.5),
        marker=dict(size=6),
    )
)

# Running MAP Estimate (Theta_MAP)
fig.add_trace(
    go.Scatter(
        x=steps,
        y=theta_map_list,
        mode="lines+markers",
        name=r"$\text{MAP Estimate } (\hat{\theta}_{\text{MAP}}^{(k)})$",
        line=dict(color="#ff7f0e", width=2, dash="dash"),
        marker=dict(size=5),
    )
)

# Reference line for true latent ability theta_true = 0.75
fig.add_trace(
    go.Scatter(
        x=[0, n_items],
        y=[theta_true, theta_true],
        mode="lines",
        name=r"$\text{True Ability } (\theta_{\text{true}} = 0.75)$",
        line=dict(color="red", width=2, dash="dot"),
    )
)

# Layout customization
fig.update_layout(
    title=dict(
        text=(
            "<b>Progression of Latent Ability Estimators</b><br><sup>Tracking"
            r" $\hat{\theta}_{\text{Bayes}}$ and $\hat{\theta}_{\text{MAP}}$"
            r" against $\theta_{\text{true}} = 0.75$ over 20 items</sup>"
        ),
        x=0.5,
    ),
    xaxis=dict(title="Item Sequence Step ($k$)", dtick=1),
    yaxis=dict(title=r"Latent Ability Estimate ($\theta$)", range=[-1.0, 2.0]),
    template="plotly_white",
    width=850,
    height=520,
    legend=dict(x=0.02, y=0.98, bordercolor="Black", borderwidth=1),
)

fig.show()
```

1. Distance Trajectory Relative to $\theta_{\text{true}}$Initial Bias (Step 0): At $k=0$, both estimators start anchored at $0.0$ due to the uninformative standard normal prior $\mathcal{N}(0,1)$. The initial distance $\vert{}\hat{\theta} - \theta_{\text{true}}\vert{}$ is relatively large ($\vert{}0.0 - 0.75\vert{} = 0.75$).Sequential Convergence ($k \to 20$): As $k$ increases, the accumulated likelihood from each user response dominates the initial prior. Consequently, the distance between both estimators ($\hat{\theta}_{\text{Bayes}}^{(k)}$ and $\hat{\theta}_{\text{MAP}}^{(k)}$) and the true latent ability $\theta_{\text{true}}$ decreases systematically, oscillating tightly around $0.75$.
2. Implications for Platform Measurement ConfidenceVariance Reduction: Every additional item response adds Fisher information ($I_k(\theta) \propto a_k^2 \cdot p_k(1-p_k)$) to the posterior distribution, squeezing its variance ($\sigma^2 \to 0$).Shrinkage of Estimator Divergence: Early in the test sequence ($k < 5$), discrete sampling fluctuations cause temporary gap divergence between the mode ($\hat{\theta}_{\text{MAP}}$) and expectation ($\hat{\theta}_{\text{Bayes}}$). As $k$ increases, the posterior distribution becomes highly symmetric and sharp (Gaussian-like by the Bernstein–von Mises theorem), making $\hat{\theta}_{\text{Bayes}}$ and $\hat{\theta}_{\text{MAP}}$ virtually identical.Platform Trust: The platform gains higher statistical confidence (narrower error bounds/credible intervals) with every item presented, allowing it to deliver dynamic assessment decisions with high accuracy.
